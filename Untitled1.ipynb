```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the NMR dataset
file_path = '/projects/cc/se_users/knlr326/1_NMR_project/2_Notebooks/MultiModalSpectralTransformer_cleaned/data/ZINK_dataset/ML_NMR_5M_XL_1H_comb_train_V8.csv'

try:
    # Read the CSV file
    df = pd.read_csv(file_path)
    
    print("=== DATASET OVERVIEW ===")
    print(f"Number of rows (items): {len(df)}")
    print(f"Number of columns: {len(df.columns)}")
    print(f"Dataset shape: {df.shape}")
    
    print("\n=== COLUMN INFORMATION ===")
    print("Column names:")
    for i, col in enumerate(df.columns):
        print(f"{i+1:2d}. {col}")
    
    print("\n=== DATA TYPES ===")
    print(df.dtypes)
    
    print("\n=== FIRST FEW ROWS ===")
    print(df.head())
    
    print("\n=== BASIC STATISTICS ===")
    print(df.describe())
    
    print("\n=== MISSING VALUES ===")
    missing_values = df.isnull().sum()
    if missing_values.sum() > 0:
        print("Columns with missing values:")
        print(missing_values[missing_values > 0])
    else:
        print("No missing values found")
    
    print("\n=== MEMORY USAGE ===")
    print(f"Memory usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    
    # If there are specific NMR-related columns, analyze them
    nmr_columns = [col for col in df.columns if any(keyword in col.lower() 
                   for keyword in ['nmr', 'chemical', 'shift', 'ppm', 'spectrum', 'peak'])]
    
    if nmr_columns:
        print(f"\n=== NMR-RELATED COLUMNS ({len(nmr_columns)}) ===")
        for col in nmr_columns:
            print(f"- {col}")
    
    # Check for SMILES or molecular structure columns
    structure_columns = [col for col in df.columns if any(keyword in col.lower() 
                        for keyword in ['smiles', 'mol', 'structure', 'inchi'])]
    
    if structure_columns:
        print(f"\n=== MOLECULAR STRUCTURE COLUMNS ({len(structure_columns)}) ===")
        for col in structure_columns:
            print(f"- {col}")
    
    print("\n=== SAMPLE DATA ===")
    print("Random sample of 3 rows:")
    print(df.sample(3))

except FileNotFoundError:
    print(f"Error: File not found at {file_path}")
    print("Please check the file path or upload the file to the current directory")
    
except Exception as e:
    print(f"Error reading file: {str(e)}")

# Alternative function to analyze uploaded CSV
def analyze_csv_file(filename):
    """
    Function to analyze any CSV file in the current directory
    """
    try:
        df = pd.read_csv(filename)
        
        print(f"=== ANALYSIS OF {filename} ===")
        print(f"Total number of items: {len(df):,}")
        print(f"Number of features: {len(df.columns)}")
        print(f"Dataset dimensions: {df.shape}")
        
        # Show column info
        print(f"\nColumns ({len(df.columns)}):")
        for i, col in enumerate(df.columns[:10]):  # Show first 10 columns
            print(f"  {i+1}. {col}")
        if len(df.columns) > 10:
            print(f"  ... and {len(df.columns) - 10} more columns")
        
        return df
        
    except Exception as e:
        print(f"Error: {str(e)}")
        return None

# Example usage if file is in current directory:
df = analyze_csv_file('ML_NMR_5M_XL_1H_comb_train_V8.csv')
```
